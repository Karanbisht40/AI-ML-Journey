```python
import numpy as np 

arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
Q1 = np.percentile(arr, 25) #25 %
Q3 = np.percentile(arr,75)  #75 %
```

```python
#IQR
IQR = Q3 -Q1
IQR
```

```python
#upper and lower fence
UF = Q3 +1.5*IQR
LF = Q1 - 1.5 *IQR
UF
```

```python
l =[]
for i in arr:
    if i>= LF and i<=UF:
        l.append(i)
print(l)
arr2 = np.array(l)
```

```python
print(arr)
print(arr2)
```

```python
import seaborn as sns

sns.boxplot(x = arr2)
# sns.boxplot(x = arr)
```

```python

#z test
import numpy as np 
from scipy.stats import norm 
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
population_mean = 170
population_std = 3
samp_mean = np.mean(sample)
n = len(sample)
```

```python
z_score =(samp_mean - population_mean)/(population_std/np.sqrt(n))
z_score
```

```python
p_value = 2 * (1 - norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#two sample test
#t test 
import numpy as np
from scipy  import stats
```

```python
groupA= [85,88,90,92,87,85,89,91,86,88]
groupB =[82,84,80,83,81,79,78,85,84,83]
```

```python
t_stats, p_value = stats.ttest_ind(groupA,groupB,equal_var=False)
```

```python
t_stats
```

```python
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
#chi sqauare test
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import chi2_contingency
```

```python
 #annova test
import pandas as pd
import seaborn as sns
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df = df[['age', 'pclass']].dropna()
```

```python
df['pclass'].unique()
```

```python
class_1 = df[df['pclass']==1]['age']
class_2 = df[df['pclass']==2]['age']
class_3 = df[df['pclass']==3]['age']
class_1
```

```python
f_stats, p_value = f_oneway(class_1,class_2,class_3)
f_stats
```

```python
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
