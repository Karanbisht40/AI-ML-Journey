perceptron is a simple algorithm used in machine learning for binary classification tasks. It works by finding a linear decision boundary that separates two classes of data points. The perceptron updates its weights based on the input features and the corresponding labels, adjusting them to minimize classification errors.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification #generate a synthetic classification dataset
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100) 
```

```python
#this function is used to calculate the output of the perceptron
#it returns 1 if the input is greater than or equal to 0, otherwise it returns 0
def perceptron(X,y):

    X= np.insert(X,0,1,axis=1) #adding 1
    weights = np.ones(X.shape[1])  
    lr = 0.1

    for i in range(1000):
        j= np.random.randint(0,100) #selecting random student
        y_hat = fun(np.dot(X[j],weights)) #Calculate prediction
        weights = weights+lr*(y[j]-y_hat)*X[j] #update weights

    return weights[0],weights[1:] #give wo, w1,w2
```

```python
def fun(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(X,y) #training the perceptron and getting the weights
```

```python
print(coef_)
```

```python
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1]) 
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(9,7))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
