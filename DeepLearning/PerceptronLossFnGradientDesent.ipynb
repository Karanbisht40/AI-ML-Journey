this files tells about the perceptron loss function and how to implement it using gradient descent. The perceptron is a simple linear classifier that can be used for binary classification tasks. The loss function measures how well the model is performing, and gradient descent is an optimization algorithm used to minimize the loss function by adjusting the model's parameters (weights and bias).

```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=15)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
    
    w1=w2=b=1
    lr = 0.1
    
    for j in range(1000):
    
        for i in range(X.shape[0]): # check every data point

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + b #yeh line is for calculating the weighted sum

            if z*y[i] < 0: #this line is for checking the condition if the point is misclassified or not
                w1 = w1 + lr*y[i]*X[i][0] 
                w2 = w2 + lr*y[i]*X[i][1]
                b = b + lr*y[i]
            
    return w1,w2,b


           
```

```python
w1,w2,b = perceptron(X,y) # call the function and get the weights and bias
```

```python
m = -(w1/w2) #m is the slope of the line and c is the y intercept of the line.
c = -(b/w2)

print(m,c)
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c #iss line m y intercept form me hai. y = mx + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)# this line is for plotting the decision boundary
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2) #ylim is used to set the limit of y axis. 
```
