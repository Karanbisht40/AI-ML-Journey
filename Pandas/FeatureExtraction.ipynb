```python
#make a new column for episode count
def extraction_episodes(txt):
    check =False
    data =""
    for i in txt:
        if i == ")":
            check= False
            return data
        if check ==True:
             data = data + i
        if i =="(":
             check = True
```

```python
df["Episodes"]=df["Title"].apply(extraction_episodes)
 
```

```python
df["Episodes"]= df["Episodes"].str.replace(" eps", "")
```

```python
def extraction_time(txt):
    check =False
    data =""
    for i in range(len(txt)):
        if txt[i]==')':
            for j in range(i+1, i+20):
                data +=txt[j]

            return data
```

```python
df['Total time'] = df['Title'].apply(extraction_time)
df.head()
```

```python
import re

def extract_months(text):
    # Extract date range like "Apr 2009 - Jul 2010"
    match = re.search(r'([A-Za-z]{3}\s\d{4})\s*-\s*([A-Za-z]{3}\s\d{4})', str(text))

    if match:
        start = pd.to_datetime(match.group(1), format="%b %Y")
        end = pd.to_datetime(match.group(2), format="%b %Y")

        # Total months (inclusive)
        return (end.year - start.year) * 12 + (end.month - start.month) + 1

    return None
```

```python
df["Months"] = df["Title"].apply(extract_months)

df.head()
```
