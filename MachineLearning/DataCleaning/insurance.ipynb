# Feature Extraction

```python
from scipy.stats import pearsonr


# List of features to check against target
selected_features = [
    'age', 'bmi', 'children', 'is_female', 'is_smoker',
    'region_northwest', 'region_southeast', 'region_southwest',
    'bmi_cetogory_normal', 'bmi_cetogory_overweight', 'bmi_cetogory_obese','age_group_18-34','age_group_35-54','age_group_55+'
]

correlations = {
    feature: pearsonr(df[feature], df['expenses'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
categorical_cols = [
    'is_smoker',
    'is_female',
    'age_group_18-34',
    'age_group_35-54',
    'age_group_55+',
    'bmi_cetogory_normal',
    'bmi_cetogory_overweight',
    'bmi_cetogory_obese',
    'region_southeast',
    'region_northwest',
    'region_southwest'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df['charges_bin'] = pd.qcut(df['expenses'], q=4, labels=False)
chi2_results = {}

for col in categorical_cols:
    contingency = pd.crosstab(df[col], df['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python

final_df = df[['age_group_18-34','age_group_35-54','age_group_55+', 'is_female', 'bmi', 'children', 'is_smoker', 'expenses','region_southeast','bmi_cetogory_obese']]
```
