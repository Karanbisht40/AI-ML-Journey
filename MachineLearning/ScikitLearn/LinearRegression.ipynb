# Linear Regression
linear regression is a statistical method that allows us to summarize and study relationships between two or more continuous (quantitative) variables. One variable, denoted y, is regarded as the response, dependent, or outcome variable. The other variable(s), denoted X, are regarded as the predictor(s), independent, or feature variables.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
data = pd.read_csv("insurance.csv")
data.head()                                                                                                                                                                         
```

```python
updated_data = pd.read_csv("insurance_updated.csv")
updated_data
```

```python

final_df = updated_data[['age_group_18-34','age_group_35-54','age_group_55+', 'is_female', 'bmi', 'children', 'is_smoker', 'expenses','region_southeast','bmi_cetogory_obese']]
```

```python
final_df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
# Define the features (X) input and target (y) output variable
X = final_df.drop('expenses', axis=1)
# Define the target variable (y)
y = final_df['expenses']
```

```python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
# Create a Linear Regression model and fit it to the training data
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Make predictions on the test set
y_pred = model.predict(X_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import r2_score

# Calculate the R-squared value to evaluate the model's performance
r2 = r2_score(y_test, y_pred) 

print("R-squared value:", r2)
```

```python
# Perform 5-fold cross-validation and calculate R² scores
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model, X, y,
    cv=5,
    scoring="r2"
)

print("CV R²:", scores.mean())
```

```python
# Calculate the adjusted R-squared value
n = X_test.shape[0]  # Number of observations in the test set
p = X_test.shape[1]  # Number of predictors (features)
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adjusted_r2

```

```python
# Calculate R² for both training and testing sets
train_r2 = r2_score(
    y_train,
    model.predict(X_train)
)

test_r2 = r2_score(
    y_test,
    model.predict(X_test)
)

print("Train R²:", train_r2)
print("Test R²:", test_r2)
```
