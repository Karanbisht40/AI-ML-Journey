heatMap

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset('flights')
flights
```

```python
tips= sns.load_dataset('tips')
tips
```

```python
tipscorr = tips[['total_bill', 'tip','size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr(),annot=True,cmap='Blues',linecolor='black')
```

cluster map


```python
sns.clustermap(tipscorr.corr(),annot=True,cmap='Blues',linecolor='black', z_score=1)
```

pivot table heat map

```python
pvtflights=  flights.pivot_table(index='month',columns='year',values='passengers')
pvtflights
```

```python
sns.heatmap(pvtflights)
```

Regression Plot

```python
tips = sns.load_dataset('tips')
sns.lmplot(data=tips, x='total_bill', y='tip')
```

```python
sns.lmplot(data=tips, x='total_bill', y='tip',hue='sex',palette='Set1',markers=['o','v'], scatter_kws={'s':50})
```
