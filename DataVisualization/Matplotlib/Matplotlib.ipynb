```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,10,9)
x
y =x**2
y
```

```python
plt.plot(x,y)
plt.title("Example")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

SubPlot

```python
plt.subplot(2,2,1)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x**2,y)

plt.subplot(2,2,3)
plt.plot(x,y**2)

plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4]
y = [10,20,15,30]

plt.figure(figsize=(8,5), dpi=150)

plt.plot(
    x,
    y,
    color="blue",
    linestyle="--",
    linewidth=2,
    marker="D",
    markersize=8,
    markerfacecolor="red",
    label="Sales"
)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
plt.legend()

plt.savefig("sales.png", dpi=300)

plt.show()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2,2)

ax[0,0].plot([1,2,3],[2,4,6])
ax[0,1].plot([1,2,3],[3,2,5])
ax[1,0].plot([1,2,3],[5,6,4])
ax[1,1].plot([1,2,3],[7,8,9])

plt.tight_layout()
plt.show()
```
