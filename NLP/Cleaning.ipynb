```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('train.txt', sep=';', header=None, names=[ 'text','emotion'])
```

```python
df.head()
```

```python
df.isnull().sum() 
```

```python
#convert the emotions into numbers
unique_emotion=df['emotion'].unique()
emotion_number={} #a empty dictionary to store
i=0
for emo in unique_emotion:
    emotion_number[emo]=i
    i+=1

df['emotion']=df['emotion'].map(emotion_number)
```

```python
df
```

```python
df['text'] = df['text'].str.lower() #convert the text into lower case
```

```python
df.head()
```

```python
#remove punctuation
import string
def remove_punctuation(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punctuation)
```

```python
df
```

```python
# remove numbers
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new + i
    return new

df['text']=df['text'].apply(remove_numbers)
```

remove emojis

```python


def remove_emojis(txt):
   new = ""
   for i in txt:
        if i.isascii():
            new = new + i
   return new

df['text']=df['text'].apply(remove_emojis)
```

Remove Stopwords

```python
import nltk
from  nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
# we set the stopwords in english 
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
    words = txt.split()
    cleaned =[]
    for i in words:
        if i not in stop_words: # jb i stopwords me nahi hoga to usko cleaned me append krdo
            cleaned.append(i)
    
    return " ".join(cleaned) 
```

```python
df['text']=df['text'].apply(remove)
```
