# Named Entity Recognition
it is used to locate and classify named entities in text into predefined categories such as the names of persons, organizations, locations, expressions of times, quantities, monetary values, percentages, etc.

```python
sentence="The Eiffel Tower was built from 1887 to 1889 by Gustave Eiffel, whose company specialized in building metal frameworks and structures."
```

```python
import nltk
words=nltk.word_tokenize(sentence) #convert sentence into words
```

```python
tag_elements=nltk.pos_tag(words) # this will give us the part of speech for each word in the sentence
tag_elements
```

```python
from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

text = "Elon Musk founded SpaceX in California."

words = word_tokenize(text)

pos_tags = pos_tag(words)

named_entities = ne_chunk(pos_tags)# ne_chunk hme name entities dega

print(named_entities)
```
