```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
corpus= """Hello welcome to the world of Natural Language Processing.
i am learning NLP ! become a NLP expert in 6 months.
"""
```

```python
print(corpus)
```

```python
# tokenize paragraph into sentences
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
documents
```

```python
type(documents)
```

```python
#tokenize sentences into words
from nltk.tokenize import word_tokenize

word_tokenize(corpus)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus) #this will also tokenize the punctuations as separate tokens
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus) 
# this tokenizer will not tokenize the punctuations as separate tokens and it will also handle the contractions like "don't" into "do" and "n't" #full stop
```
