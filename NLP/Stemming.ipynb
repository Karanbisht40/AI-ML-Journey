# Stemming (text processing technique)
PorterStemmer


```python
words = ["eating","eats","eaten","writing", "writes","programming","programs","history","finally","finalized"]
```

```python
from nltk.stem import PorterStemmer

stemming = PorterStemmer()

for word in words:
    print(word +"--->"+ stemming.stem(word))
```

```python
stemming.stem("congratulations") #meaningless words 
```

# RegexpStemmer
it is a more flexible stemming technique that allows you to define custom rules for removing suffixes from words.

```python
from nltk.stem import RegexpStemmer

reg_stremmer = RegexpStemmer('ing$|s$|e$|ed$', min=4)
reg_stremmer.stem("eating")
reg_stremmer.stem("ingeating")
```

# SnowBall Stemmer
it is an improvement over the Porter Stemmer and supports multiple languages. It uses a more sophisticated algorithm to handle different word forms and is generally more accurate than the Porter Stemmer.

```python
from nltk.stem import SnowballStemmer

snowball_stemmer = SnowballStemmer("english")

for word in words:
    print(word +"--->"+ snowball_stemmer.stem(word))
```

```python
stemming.stem('fairly'),stemming.stem('sportingly') #stemming
```

```python
snowball_stemmer.stem('fairly'),snowball_stemmer.stem('sportingly') #snowball
```
