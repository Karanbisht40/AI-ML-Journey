# Lemmatization (Text processing technique)
wordNetLemmatizer is a more advanced technique than stemming. It reduces words to their base or dictionary form (lemma) based on their meaning and context. 

```python
words = ["eating","eats","eaten","writing", "writes","programming","programs","history","finally","finalized"]
```

```python
import nltk
nltk.download('wordnet') 
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
for word in words:
    print(word +"--->"+ lemmatizer.lemmatize(word, pos='v')) #pos='v' indicates that the word is a verb
```

```python
lemmatizer.lemmatize('fairly', pos='r'), lemmatizer.lemmatize('sportingly', pos='v') 
```
